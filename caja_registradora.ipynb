# 🎯 Fase 1: El problema

Un negocio pequeño necesita **cobrar rápido y sin errores**. Hacer las cuentas a mano o con calculadora puede causar:

- ❌ Errores al sumar
- ❌ Descuentos mal aplicados
- ❌ Olvidar el impuesto
- ❌ Errores al dar el cambio

✅ **Nuestra solución:** un programa que haga todos los cálculos automáticamente.

- [ ] Rechazar precios o cantidades menores o iguales a 0
- [ ] Aplicar descuento según el monto de la compra
- [ ] Dar un descuento extra a clientes frecuentes
- [ ] Calcular el ITBIS (impuesto)
- [ ] Mostrar una factura clara
- [ ] Calcular el cambio (o lo que falta si no alcanza)

### 🏷️ Reglas de descuento

| RD$ 2,000 o más | 5% |
| RD$ 5,000 o más | 10% |
| Cliente frecuente (si ya tiene descuento) | **+2% extra** |

# 💻 Fase 4: Código paso a paso

Cada paso tiene una **explicación** y luego el **código**. Ejecuta las celdas en orden.

## ⚙️ Paso 1: Configuración inicial

Creamos una **constante**: un valor que no cambia durante el programa. Aquí guardamos la tasa del impuesto (ITBIS).

> 📝 Por convención, las constantes se escriben en **MAYÚSCULAS**.
> Si la tasa cambia algún día, solo modificas esta línea y todo el programa se actualiza.

````python
ITBIS = 0.18  # 18% expresado como decimal

print("=== CAJA REGISTRADORA ===")
````

## 👤 Paso 2: Datos del cliente

Usamos `input()` para **preguntar** al usuario. Importante:

| Función | Qué hace | Ejemplo |
|---|---|---|
| `input()` | Siempre devuelve **texto** (`str`) | `"5"` |
| `int()` | Convierte texto a número entero | `5` |
| `float()` | Convierte texto a número decimal | `5.50` |

Por eso envolvemos con `int(...)` la cantidad de productos: **necesitamos un número**, no texto.

También creamos `subtotal = 0.0`, una variable **acumuladora**: empieza en cero y le iremos sumando cada producto.

````python
nombre_cliente = input("Nombre del cliente: ")
cantidad_productos = int(input("¿Cuántos productos diferentes compra? "))

subtotal = 0.0  # aquí se irá acumulando el total de la compra
````

## 🔁 Paso 3: Registrar los productos (el loop)

Este es el **corazón del programa**. Usamos `for` con `range()` para repetir las mismas preguntas por cada producto.

### Cómo funciona `range(1, cantidad_productos + 1)`

Si el cliente compra **3** productos:

| Vuelta | Valor de `i` |
|---|---|
| 1ª | 1 |
| 2ª | 2 |
| 3ª | 3 |

> ❓ **¿Por qué `+ 1`?** Porque `range` **no incluye** el último número. `range(1, 4)` da 1, 2 y 3.

### La validación con `or`

```python
if precio <= 0 or cantidad <= 0:
```
Significa: *"si el precio es inválido **O** la cantidad es inválida"*. Con que **una** sea verdadera, se rechaza el producto.

### El acumulador

```python
subtotal = subtotal + total_producto
```
Toma lo que ya había en `subtotal` y le suma el nuevo producto.

````python
for i in range(1, cantidad_productos + 1):
    print("\nProducto", i)
    nombre = input("Nombre: ")
    precio = float(input("Precio unitario (RD$): "))
    cantidad = int(input("Cantidad: "))

    if precio <= 0 or cantidad <= 0:
        print("Datos inválidos. Este producto no se cuenta.")
    else:
        total_producto = precio * cantidad
        subtotal = subtotal + total_producto
        print(nombre, "x", cantidad, "= RD$", total_producto)
````

## 🏷️ Paso 4: Calcular el porcentaje de descuento

Aquí usamos **`if / elif / else`**: Python revisa las condiciones **en orden** y ejecuta solo la primera que sea verdadera.

```
¿subtotal >= 5000?  → sí → 10%
        ↓ no
¿subtotal >= 2000?  → sí → 5%
        ↓ no
        → 0%
```

> ⚠️ **El orden importa.** Si pusiéramos primero `>= 2000`, una compra de 6000 recibiría solo 5%, porque ya cumple esa condición.

### Cliente frecuente con `and`

```python
if es_frecuente and descuento_porcentaje > 0:
```
`and` exige que **las dos** condiciones sean verdaderas: ser cliente frecuente **Y** ya tener algún descuento.

> 🧠 `input(...) == "si"` compara la respuesta con la palabra "si". El resultado es `True` o `False` (un valor **booleano**).

````python
es_frecuente = input("¿Es cliente frecuente? (si/no): ") == "si"

if subtotal >= 5000:
    descuento_porcentaje = 0.10
elif subtotal >= 2000:
    descuento_porcentaje = 0.05
else:
    descuento_porcentaje = 0.0

# Cliente frecuente: 2% extra, pero solo si ya tiene algún descuento
if es_frecuente and descuento_porcentaje > 0:
    descuento_porcentaje = descuento_porcentaje + 0.02
````

## 🧮 Paso 5: Hacer los cálculos

Aplicamos las fórmulas en este orden (el orden **sí** importa):

| Cálculo | Fórmula |
|---|---|
| Descuento en pesos | `subtotal * descuento_porcentaje` |
| Subtotal con descuento | `subtotal - descuento` |
| ITBIS | `subtotal_con_descuento * ITBIS` |
| **Total a pagar** | `subtotal_con_descuento + impuesto` |

> 💡 El impuesto se calcula **después** del descuento. Es más justo para el cliente.

````python
descuento = subtotal * descuento_porcentaje
subtotal_con_descuento = subtotal - descuento
impuesto = subtotal_con_descuento * ITBIS
total = subtotal_con_descuento + impuesto
````

## 🧾 Paso 6: Mostrar la factura

Usamos `print()` para mostrar los resultados y `round(numero, 2)` para **redondear a 2 decimales** (sin esto podrían salir números como `871.2000000001`).

````python
print("\n=========== FACTURA ===========")
print("Cliente:", nombre_cliente)
print("Subtotal:           RD$", round(subtotal, 2))
print("Descuento:         -RD$", round(descuento, 2))
print("ITBIS:              RD$", round(impuesto, 2))
print("TOTAL A PAGAR:      RD$", round(total, 2))
print("===============================")
````

## 💵 Paso 7: Pago y cambio

Primero verificamos que **haya algo que cobrar** (`total > 0`). Luego comparamos el pago con el total:

- Si `pago >= total` → damos el cambio (`pago - total`)
- Si no → mostramos cuánto falta (`total - pago`)

> 🧠 Fíjate que hay un `if` **dentro** de otro `if`. Se llama **condicional anidado**.

````python
if total > 0:
    pago = float(input("\n¿Con cuánto paga el cliente? RD$ "))
    if pago >= total:
        print("Cambio: RD$", round(pago - total, 2))
        print("¡Gracias por su compra!")
    else:
        print("Pago insuficiente. Faltan RD$", round(total - pago, 2))
else:
    print("\nNo hay nada que cobrar.")
````

### 📟 Así se ve el resultado (ejemplo)

Con *Arroz 100 x 30*, *Pollo 250 x 10*, cliente frecuente, pagando con 6000:

